# **Project Summary -**

This project undertook a comprehensive analysis of Zomato's restaurant and review data to extract actionable business insights through data wrangling, exploratory analysis, and the implementation of unsupervised machine learning. The primary objective was to segment restaurants into distinct clusters to better understand the market landscape and to analyze customer sentiment to gauge user satisfaction.

The initial phase focused on robust data preparation. Two separate datasets—one containing restaurant metadata and the other containing user reviews—were loaded and meticulously cleaned. Key data wrangling steps included converting the 'Cost' column from a text-based object to a numerical format, handling missing values by imputing 'Cuisines' with an "Unknown" category, and dropping incomplete review entries. The most critical step was feature engineering, where the extensive review data was aggregated to calculate each restaurant's average_rating and total review_count. These newly engineered features were then merged with the restaurant metadata to create a single, unified dataset, providing a 360-degree view of each establishment and setting the foundation for all subsequent analysis.

Exploratory Data Analysis revealed several key trends. Visualizations showed that the market is heavily dominated by a few popular cuisines, primarily North Indian and Chinese, indicating potential saturation in these areas. Analysis of cost and rating distributions highlighted a market skewed towards lower-cost establishments. Hypothesis testing was employed to statistically validate these observations, confirming a significant, albeit weak, positive correlation between a restaurant's cost and its average rating.

The core of the project was the implementation of a K-Means clustering algorithm to segment restaurants based on their cost, average rating, and review volume. To determine the optimal number of clusters, both the Elbow Method and Silhouette Analysis were utilized, which collectively pointed to k=4 as the most effective configuration. This resulted in the creation of four distinct and interpretable restaurant personas: a "Budget-Friendly" segment with low costs and moderate ratings; a "Popular Mid-Range" group with high review counts; a "Premium Dining" cluster with high costs and high ratings; and a "High-Value Gems" segment with moderate costs and high ratings. By analyzing the centroids of these clusters, we determined that Cost and review_count were the most influential features in differentiating these groups.

In parallel, a sentiment analysis was performed on the review text. By calculating a sentiment polarity score for each review, we found a strong positive correlation between the sentiment expressed in the text and the star rating awarded by the user. This finding validates the integrity of Zomato's rating system, confirming that ratings are a reliable reflection of customer satisfaction.

In conclusion, this project successfully transformed raw data into strategic business intelligence. The identified restaurant clusters provide Zomato with a data-driven framework for targeted marketing, personalized user recommendations, and strategic partner acquisition by highlighting underserved market segments. The sentiment analysis reinforces the value of customer reviews and opens avenues for enhanced feedback systems for restaurant partners. Ultimately, the insights from this analysis empower Zomato to better understand its ecosystem, improve its platform, and drive sustainable business growth.

## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
df_rest = restaurants_df.copy()
df_rev = reviews_df.copy()
df_rest["Cost"] = df_rest["Cost"].str.replace(",","").astype(float)
df_rest["Cuisines"] = df_rest["Cuisines"].fillna('Unknown')
df_rev.dropna(subset = ['Review'], inplace =True)
df_rev['Rating'] = pd.to_numeric(df_rev['Rating'], errors='coerce')

review_summary = df_rev.groupby('Restaurant').agg(
    average_rating=('Rating', 'mean'),
    review_count=('Rating', 'count')
).reset_index()
review_summary['average_rating'] = review_summary['average_rating'].round(2)
df_final = pd.merge(df_rest, review_summary, left_on='Name', right_on='Restaurant', how='inner')
df_final = df_final.drop(columns=['Restaurant', 'Links'])
print("Data wrangling completed")
print("Final dataset shape:", df_final.shape)
print("\n--- First 5 Rows of Final Dataset ---")
print(df_final.head())
````

### What all manipulations have you done and insights you found?

In this section, I performed several key manipulations to prepare the two separate datasets for analysis:

I handled data quality issues. The Cost column in the restaurant data was converted from a text field with commas into a proper numerical format. I also managed missing values by filling empty Cuisines with the label "Unknown" and removing reviews that had no text content.

The most important manipulation was summarizing the review data. Since the goal is to analyze restaurants, I aggregated the thousands of individual reviews to create two powerful new features for each restaurant: its average_rating and its total review_count.

Finally, I merged the cleaned restaurant data with the newly created review summary. This consolidated all the relevant information into a single, unified dataset (df_final).Answer Here.

#### Chart - 10

````python
# Chart - 10 visualization code
# Ensure you have run the data wrangling and clustering code from previous steps
# to create the 'df_final' DataFrame with the 'cluster' column.

from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

# Select the features for clustering
features = ['Cost', 'average_rating']
X = df_final[features].copy()

# Handle potential infinite values in 'Cost' after division
X.replace([np.inf, -np.inf], np.nan, inplace=True)
X.dropna(inplace=True)

# It's important to scale the data before applying KMeans
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Apply KMeans clustering (choose a suitable number of clusters, e.g., 3)
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10) # Added n_init=10
df_final.loc[X.index, 'cluster'] = kmeans.fit_predict(X_scaled)


plt.figure(figsize=(12, 8))
sns.scatterplot(data=df_final, x='Cost', y='average_rating', hue='cluster', palette='viridis', s=100, alpha=0.7)

plt.title('Zomato Restaurant Clusters: Cost vs. Average Rating', fontsize=16)
plt.xlabel('Average Cost Per Person (INR)', fontsize=12)
plt.ylabel('Average Customer Rating', fontsize=12)
plt.legend(title='Restaurant Segment')
plt.show()
````

## ***5. Hypothesis Testing***

### Based on your chart experiments, define three hypothetical statements from the dataset. In the next three questions, perform hypothesis testing to obtain final conclusion about the statements through your code and statistical testing.

* Restaurants with higher ratings tend to have higher prices.
* The rating of a restaurant doesn't seem to be influenced by the number of followers its most-followed reviewer has.
* Restaurants offering a greater variety of cuisines may receive higher ratings.

### Hypothetical Statement - 1 (Restaurants with higher ratings tend to have higher prices.)

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

Null Hypothesis (H_0): There is no correlation between the average rating of a restaurant and its per-person cost. The population correlation coefficient is zero (
rho=0).

Alternate Hypothesis (H_a): There is a positive correlation between the average rating of a restaurant and its per-person cost. The population correlation coefficient is greater than zero (
rho0).

#### 2. Perform an appropriate statistical test.

````python
from scipy.stats import pearsonr

# Drop any potential NaN values to ensure the test runs correctly
test_df = df_final[['Cost', 'average_rating']].dropna()

# Perform the Pearson correlation test
correlation_coefficient, p_value = pearsonr(test_df['Cost'], test_df['average_rating'])

print(f"Correlation Coefficient: {correlation_coefficient:.4f}")
print(f"P-value: {p_value:.4f}")

# Conclusion based on the p-value
alpha = 0.05
if p_value < alpha:
    print("\nConclusion: We reject the null hypothesis.")
    print("There is a statistically significant positive correlation between cost and average rating.")
else:
    print("\nConclusion: We fail to reject the null hypothesis.")
    print("There is no statistically significant correlation between cost and average rating.")
````

##### Which statistical test have you done to obtain P-Value?

I performed a Pearson correlation test.

##### Why did you choose the specific statistical test?

I chose the Pearson correlation test because it is the standard method for measuring the strength and direction of a linear relationship between two continuous variables (Cost and average_rating).

### Hypothetical Statement - 2 (The rating of a restaurant doesn't seem to be influenced by the number of followers its most-followed reviewer has)

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

Null Hypothesis (H_0): There is no correlation between a restaurant's average rating and the follower count of its most-followed reviewer. The population correlation coefficient is zero (
rho=0).

Alternate Hypothesis (H_a): There is a correlation between a restaurant's average rating and the follower count of its most-followed reviewer. The population correlation coefficient is not zero (
rho
neq0).

#### 2. Perform an appropriate statistical test.

````python
from scipy.stats import pearsonr
import numpy as np

# --- Data Preparation ---
# 1. Extract follower count from the metadata
followers = reviews_df['Metadata'].str.extract(r'(\d+)\s+Followers').astype(float)
reviewers_with_followers = reviews_df[['Restaurant']].copy()
reviewers_with_followers['followers'] = followers
reviewers_with_followers.dropna(inplace=True)

# 2. For each restaurant, find the maximum follower count among its reviewers
max_followers_per_rest = reviewers_with_followers.groupby('Restaurant')['followers'].max().reset_index()

# 3. Merge this with our final dataframe which has average_rating
hypothesis_df = pd.merge(df_final, max_followers_per_rest, left_on='Name', right_on='Restaurant', how='inner')

# --- Perform the Statistical Test ---
# Drop any potential NaN values
test_df = hypothesis_df[['followers', 'average_rating']].dropna()

correlation_coefficient, p_value = pearsonr(test_df['followers'], test_df['average_rating'])

print(f"Correlation Coefficient: {correlation_coefficient:.4f}")
print(f"P-value: {p_value:.4f}")

# Conclusion based on the p-value
alpha = 0.05
if p_value < alpha:
    print("\nConclusion: We reject the null hypothesis.")
    print("There is a statistically significant correlation between a restaurant's rating and its top reviewer's follower count.")
else:
    print("\nConclusion: We fail to reject the null hypothesis.")
    print("There is no statistically significant correlation.")
````

This test precisely measures if these two variables move in relation to each other.

#### 2. Perform an appropriate statistical test.

````python
from scipy.stats import pearsonr

# --- Data Preparation ---
# Create a new column for the count of cuisines offered by each restaurant
hypothesis_df = df_final.copy()
hypothesis_df['cuisine_count'] = hypothesis_df['Cuisines'].str.split(',').str.len()

# --- Perform the Statistical Test ---
# Drop any potential NaN values
test_df = hypothesis_df[['cuisine_count', 'average_rating']].dropna()

correlation_coefficient, p_value_2_tailed = pearsonr(test_df['cuisine_count'], test_df['average_rating'])

# For a one-tailed test (since we're testing for a POSITIVE correlation)
p_value_1_tailed = p_value_2_tailed / 2

print(f"Correlation Coefficient: {correlation_coefficient:.4f}")
print(f"P-value (One-tailed): {p_value_1_tailed:.4f}")

# Conclusion based on the p-value
alpha = 0.05
if p_value_1_tailed < alpha and correlation_coefficient > 0:
    print("\nConclusion: We reject the null hypothesis.")
    print("There is a statistically significant positive correlation between cuisine count and average rating.")
else:
    print("\nConclusion: We fail to reject the null hypothesis.")
    print("There is no statistically significant positive correlation.")
````

````python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# Use the scaled features from the data preparation steps
# scaled_features = scaler.fit_transform(features)

# --- ML Model - 1 Implementation ---
# Initialize the K-Means model
kmeans = KMeans(n_clusters=4, init='k-means++', random_state=42, n_init=10)

# --- Fit the Algorithm ---
# Select the numerical columns from the scaled_df for clustering
numerical_cols_scaled = scaled_df.select_dtypes(include=np.number).columns.tolist()
# Remove the 'cluster' and 'Cost_capped' column if they are present, as they are not original features for clustering
if 'cluster' in numerical_cols_scaled:
    numerical_cols_scaled.remove('cluster')
if 'Cost_capped' in numerical_cols_scaled:
    numerical_cols_scaled.remove('Cost_capped')

kmeans.fit(scaled_df[numerical_cols_scaled])


# --- Predict on the model ---
# For clustering, "predicting" means assigning a cluster label to each data point.
cluster_labels = kmeans.labels_

# Add the cluster labels back to our main dataframe
df_final.loc[scaled_df.index, 'cluster'] = cluster_labels

print("K-Means model has been fitted and cluster labels are assigned.")
````

````python
# Visualizing evaluation Metric Score chart
# --- Visualizing evaluation Metric Score chart ---
# Calculate the Silhouette Score for our model
# Select the numerical columns from the scaled_df for clustering
numerical_cols_scaled = scaled_df.select_dtypes(include=np.number).columns.tolist()
# Remove the 'cluster' and 'Cost_capped' column if they are present, as they are not original features for clustering
if 'cluster' in numerical_cols_scaled:
    numerical_cols_scaled.remove('cluster')
if 'Cost_capped' in numerical_cols_scaled:
    numerical_cols_scaled.remove('Cost_capped')

score = silhouette_score(scaled_df[numerical_cols_scaled], cluster_labels)

# Create a bar chart for the score
plt.figure(figsize=(6, 4))
sns.barplot(x=['K-Means (k=4)'], y=[score])
plt.title('Silhouette Score for K-Means Model')
plt.ylabel('Score')
plt.ylim(0, 1) # Set y-axis limits for better comparison
plt.show()

print(f"The Silhouette Score for k=4 is: {score:.4f}")
````

#### 2. Cross- Validation & Hyperparameter Tuning

````python
# ML Model - 1 Implementation with hyperparameter optimization techniques (i.e., GridSearch CV, RandomSearch CV, Bayesian Optimization etc.)
# --- ML Model - 1 Implementation with hyperparameter optimization ---
# We test different values of k (from 2 to 10) and find the one with the best Silhouette Score.
silhouette_scores = []
k_values = range(2, 11)

# Select the numerical columns from the scaled_df for clustering
numerical_cols_scaled = scaled_df.select_dtypes(include=np.number).columns.tolist()
# Remove the 'cluster' and 'Cost_capped' column if they are present, as they are not original features for clustering
if 'cluster' in numerical_cols_scaled:
    numerical_cols_scaled.remove('cluster')
if 'Cost_capped' in numerical_cols_scaled:
    numerical_cols_scaled.remove('Cost_capped')

# Extract the scaled features array
scaled_features_array = scaled_df[numerical_cols_scaled].values


for k in k_values:
    # Fit the Algorithm
    kmeans_k = KMeans(n_clusters=k, init='k-means++', random_state=42, n_init=10)
    kmeans_k.fit(scaled_features_array)

    # Predict on the model
    labels_k = kmeans_k.labels_

    # Calculate silhouette score
    score_k = silhouette_score(scaled_features_array, labels_k)
    silhouette_scores.append(score_k)

# Visualize the scores to find the optimal k
plt.figure(figsize=(10, 6))
plt.plot(k_values, silhouette_scores, 'bo-')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('Silhouette Score')
plt.title('Silhouette Analysis for Optimal k')
plt.show()
````

Silhouette analysis provides a score for different values of k (the main hyperparameter), allowing us to choose the k that results in the densest and most well-separated clusters.
